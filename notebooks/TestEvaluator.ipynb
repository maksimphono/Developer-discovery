```python
from src.utils.Evaluator import Evaluator
from src.utils.CacheAdapter import EvaluationAdapterFactory

relatedAda, unrelatedAda = EvaluationAdapterFactory.createProjectsEvaluationGroups()

class Corp:
    def __getitem__(self, indexes):
        index = indexes[0]
        return {
            0 : [{"words" : ["One", "Two"], "tags" : ["1", "2"]}], 
            11 : [{"words" : ["Four"], "tags" : ["4"]}], 
            157 : [{"words" : ["mnbv"], "tags" : ["4"]}],
            1665 : [{"words" : ["Two", "Tree"], "tags" : ["2", "3"]}],
            4762 : [{"words" : ["asdf"], "tags" : ["2", "3"]}]
        }[index]

testCorpus = Corp()
def sims(a, b):
    print(f"Sim {a}, {b}")
    return int(a == b)

def model(doc):
    return {
        "['One', 'Two']" : [1,1],
        "['Four']" : [1,0],
        "['Two', 'Tree']" : [1,1],
        "['asdf']" : [1,0],
        "['mnbv']" : [1,0],
    }[str(doc["words"])]

evaluator = Evaluator(relatedAda, unrelatedAda, testCorpus, 2)
evaluator.setModel(model)
evaluator.setSimilarityCheck(sims)
evaluator.evaluate()
```
